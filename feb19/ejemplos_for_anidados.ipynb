```python
# ejemplo tablas de multiplicar
for i in range(1, 11): # inicia un bucle for que se repetirá 10 veces, con i tomando valores de 1 a 10
    for j in range(1, 11): # dentro del primer bucle, inicia otro bucle for que se repetirá 10 veces, con j tomando valores de 1 a 10
        print(f"{i} x {j} = {i * j}", end="\t") # imprime la multiplicación de i y j en formato de tabla, con un tabulador entre cada resultado
    print() # imprime una nueva línea al final de cada fila para separar las tablas de multiplicar
```

```python
import random as rd # importa el módulo random y lo asigna al alias rd para generar números aleatorios
sumaTotal = 0 # inicializa la variable sumaTotal en 0 para acumular la suma de todos los números generados en todas las filas
for i in range(10):
    # Si pones algo antes del for interno, se ejecutará solo una vez por fila, al empezar el for interno, no por cada número aleatorio
    print(f"Fila {i + 1}: ") # imprime el número de fila seguido de un espacio, sin saltar a la siguiente línea
    suma = 0 # inicializa la variable suma en 0 para acumular la suma de los números aleatorios generados en cada fila
    for j in range(10):        
        num = rd.randint(0, 9) # genera un número entero aleatorio entre 0 y 9 (inclusive)
        print(num, end="") # imprime el número generado seguido de un espacio, sin saltar a la siguiente línea
        suma += num # acumula el número generado en la variable suma
    # Si pones algo despues del for interno, se ejecutará solo una vez por fila, al acabar el for interno, no por cada número aleatorio
    print(f"\nSuma: {suma}") # imprime la suma de los números generados en cada fila seguida de una nueva línea
    sumaTotal += suma # acumula la suma de cada fila en la variable sumaTotal
print(f"Suma total: {sumaTotal}") # imprime la suma total de todos los números generados en todas las filas
```
