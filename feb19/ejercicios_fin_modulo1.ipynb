```python
# Pedimos el número al usuario
num = input("Enter an integer: ")
# Comprobamos si es palíndromo
if num == num[::-1]: # compara el número con su reverso utilizando slicing para invertir la cadena
  print(num, "is a palindrome number")
else:
  print(num, "is not a palindrome number")
```

```python
palabra = "Roma ni se conoce sin oro ni se conoce sin amor".lower().replace(" ", "") # elimina los espacios de la palabra para considerar solo las letras al verificar el palíndromo
palindromo = True # inicializa una variable para indicar si la palabra es un palíndromo
for posicion in range(len(palabra)//2): # itera sobre los índices de la palabra
    letra = palabra[posicion] # obtiene la letra en la posición actual
    letra_opuesta = palabra[-(posicion + 1)] # obtiene la letra en la posición opuesta desde el final
    if letra != letra_opuesta: # compara la letra actual con su opuesta
        print("La palabra no es un palíndromo") # si las letras no coinciden, la palabra no es un palíndromo
        palindromo = False # marca la palabra como no palíndromo
        break # sale del bucle si se encuentra una diferencia
if palindromo:
    print("La palabra es un palíndromo")
```

```python
import random as rd # Importa el módulo random y lo renombra como rd
numero_aleatorio = rd.randint(1, 100) # Genera un número aleatorio entre 1 y 100 y lo guarda en la variable numero_aleatorio
adivinado = False # Inicializa la variable adivinado en False, asumiendo que el número no ha sido adivinado hasta que se demuestre lo contrario
intentos = 0 # Inicializa la variable intentos en 0 para contar el número de intentos del usuario
while not adivinado: # Mientras el número no haya sido adivinado, repite el bucle
    numero_usuario = int(input("Introduce un número entre 1 y 100: ")) 
    intentos += 1 # Incrementa el contador de intentos en 1
    if numero_usuario == numero_aleatorio:
        print("¡Correcto! Has adivinado el número.") 
        print(f"Has necesitado {intentos} intentos para adivinar el número.")
        adivinado = True 
    elif numero_usuario > numero_aleatorio: 
        print("El número es menor.") 
    else: 
        print("El número es mayor.") 
```
