```python
# ejemplo while imprimiendo números del 1 al 10
contador = 1 # inicializa la variable contador en 1
while contador <= 10: # mientras contador sea menor o igual a 10, se ejecuta el bloque de código dentro del while
    print(contador) # imprime el valor actual de contador en cada iteración, mostrando los números del 1 al 10 en la salida
    contador += 1 # incrementa el valor de contador en 1 en cada iteración, asegurando que el bucle eventualmente termine cuando contador alcance 11
```

```python
numero_secreto = 7 # define un número secreto que el usuario debe adivinar
adivinado = False # inicializa la variable adivinado en False para controlar el bucle
while not adivinado: # mientras adivinado sea False, se ejecuta el bloque de código dentro del while
    intento = int(input("Adivina el número secreto (entre 1 y 10): ")) # solicita al usuario que introduzca un número y lo almacena en la variable intento
    if intento == numero_secreto: # verifica si el intento del usuario es igual al número secreto
        print("¡Felicidades! Has adivinado el número secreto.") # si el intento es correcto, imprime un mensaje de felicitación
        adivinado = True # establece adivinado en True para salir del bucle
    else: # si el intento no es correcto
        print("¡Inténtalo de nuevo!") # si el intento es incorrecto, imprime un mensaje de ánimo para que el usuario intente nuevamente
print("¡Gracias por jugar!") # imprime un mensaje de agradecimiento después de que el usuario adivine el número secreto y salga del bucle
```

```python
respuesta = "s" # inicializa la variable respuesta con el valor "s" para controlar el bucle
while respuesta.lower() == "s": # mientras la respuesta sea "s" (ignorando mayúsculas), se ejecuta el bloque de código dentro del while
    print("¡Hola! Este es un bucle que se repetirá mientras respondas 's'.") # imprime un mensaje de saludo en cada iteración del bucle
    respuesta = input("¿Quieres continuar? (s/n): ") # solicita al usuario que introduzca una respuesta y la almacena en la variable respuesta
    if respuesta.lower() != "s": # verifica si la respuesta del usuario no es "s" (ignorando mayúsculas)
        break # sale del bucle si la respuesta no es "s"
    print("Continuando con la partida...") # imprime un mensaje indicando que se está continuando con la partida después de que el usuario responda "s"
print("¡Gracias por jugar!") # imprime un mensaje de agradecimiento después de que el usuario
```

```python
# Ejemplo con continue
contador = 0 # inicializa la variable contador en 0
while contador < 10: # mientras contador sea menor que 10, se ejecuta el bloque de código dentro del while
    contador += 1 # incrementa el valor de contador en 1 en cada iteración
    if contador % 2 == 0: # verifica si contador es un número par
        continue # si contador es par, salta el resto del código en esta iteración y continúa con la siguiente iteración del bucle
    print(contador) # imprime el valor actual de contador solo si es impar, mostrando los números impares del 1 al 9 en la salida
```
