```python
# Ejemplos de la librería random
import random as rd
# Ejemplo 1: Generar un número aleatorio entre 1 y 10
numero_aleatorio = rd.randint(1, 10) # genera un número entero aleatorio entre 1 y 10 (inclusive) y lo almacena en la variable numero_aleatorio
print(f"Número aleatorio entre 1 y 10: {numero_aleatorio}")
# Ejemplo 2: Seleccionar un elemento aleatorio de una lista
colores = ["rojo", "verde", "azul", "amarillo"] # define una lista de colores
color_aleatorio = rd.choice(colores) # selecciona un elemento aleatorio de la lista colores y lo almacena en la variable color_aleatorio
print(f"Color aleatorio seleccionado: {color_aleatorio}")
# Ejemplo 3: Mezclar una lista de números
numeros = [1, 2, 3, 4, 5] # define una lista de números
rd.shuffle(numeros) # mezcla aleatoriamente los elementos de la lista numeros
print(f"Números mezclados: {numeros}")
# Ejemplo 4: Generar un número aleatorio con distribución normal
media = 0 # define la media de la distribución normal
desviacion_estandar = 1 # define la desviación estándar de la distribución normal
numero_normal = rd.gauss(media, desviacion_estandar) # genera un número aleatorio con distribución normal utilizando la media y desviación estándar especificadas, y lo almacena en la variable numero_normal
print(f"Número aleatorio con distribución normal (media={media}, desviación estándar={desviacion_estandar}): {numero_normal:.2f}")
```

```python
# Un programa que simula una espera aleatoria entre 1 y 5 segundos antes de mostrar un mensaje
import time
import random as rd
espera = rd.randint(1, 5) # genera un número entero aleatorio entre 1 y 5 (inclusive) y lo almacena en la variable esperaimiento
print(f"Esperando {espera} segundos...") # muestra un mensaje indicando el tiempo de espera
time.sleep(espera) # hace que el programa espere durante el número de segundos especificado
print("¡Tiempo de espera terminado!") # muestra un mensaje indicando que el tiempo de espera ha terminado
```
