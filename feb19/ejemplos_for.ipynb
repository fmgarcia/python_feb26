```python
suma = 0 # inicializa la variable suma en 0
for i in range(1, 11): # itera desde i=1 hasta i=10 (inclusive)
    suma += i # agrega el valor actual de i a la variable suma en cada iteración
print(suma) # imprime el valor final de suma después de la iteración, mostrando la suma de los números del 1 al 10 en la salida
```

```python
print(sum(range(1, 11))) # utiliza la función sum() para calcular la suma de los números del 1 al 10, mostrando el resultado en la salida
```

```python
# Quiero sumar todos los números pares del 1 al 100
print(sum(range(2, 101, 2))) # utiliza la función sum() para calcular la suma de los números pares del 1 al 100, mostrando el resultado en la salida
```

```python
suma = 0 # inicializa la variable suma en 0
for i in range(1, 101): # itera desde i=1 hasta i=100 (inclusive)
    if i % 2 == 0: # verifica si el número actual i es par (es decir, si el resto de la división de i entre 2 es igual a 0)
        suma += i # si i es par, agrega su valor a la variable suma
print(suma) # imprime el valor final de suma después de la iteración, mostrando la suma de los números pares del 1 al 100 en la salida
```

```python
# Ejemplo de factorial con bucle for
numero = int(input("Introduce un número para calcular su factorial: ")) # solicita al usuario que introduzca un número y lo almacena en la variable "numero"
resultado = 1 # inicializa la variable resultado en 1
for i in range(1, numero + 1): # itera desde i=1 hasta i=numero (inclusive)
    resultado *= i # multiplica el valor actual de resultado por i en cada iteración, calculando el factorial del número
print(f"El factorial de {numero} es {resultado}") # imprime el resultado del factorial utilizando una f-string para mostrar el número y su factorial en la salida
```

```python
# formatos de impresión de números en print()
numero = 3.141592653589793
numero2 = 123456.789
print(f"El número pi con 2 decimales es: {numero:.2f}") # utiliza una f-string para imprimir el número pi con 2 decimales, mostrando "El número pi con 2 decimales es: 3.14" en la salida
print(f"El número pi con 4 decimales es: {numero:.4f}") # utiliza una f-string para imprimir el número pi con 4 decimales, mostrando "El número pi con 4 decimales es: 3.1416" en la salida
print(f"El número2 con notación científica es: {numero2:.2e}") # utiliza una f-string para imprimir el número2 en notación científica con 2 decimales, mostrando "El número2 con notación científica es: 1.23e+05" en la salida
print(f"El número2 con notación científica es: {numero2:.4e}") # utiliza una f-string para imprimir el número2 en notación científica con 4 decimales, mostrando "El número2 con notación científica es: 1.2346e+05" en la salida
print(f"El número2 con formato de miles es: {numero2:,.2f}") # utiliza una f-string para imprimir el número2 con formato de miles y 2 decimales, mostrando "El número2 con formato de miles es: 123,456.79" en la salida
print(f"El número2 con X números enteros y 2 decimales es: {numero2:10.2f}") # utiliza una f-string para imprimir el número2 con un ancho total de 10 caracteres, incluyendo 5 números enteros y 2 decimales, mostrando "El número2 con 5 números enteros y 2 decimales es: 123456.79" en la salida
print(f"El número2 con X números enteros y 2 decimales es: {numero2:5.2f}")
print(f"El número2 con X números enteros y 2 decimales es dejando 20 espacios alineados a la derecha: {numero2:15.2f}") #
```

```python
# Ejercicio pág 489
n = int(input("Ingrese el tamaño de la matriz: "))
for i in range(n * n):
    fila = i // n
    col = i % n
    if fila % 2 == 0:
        valor = fila * n + col + 1
    else:
        valor = fila * n + (n - col)
    print(f"{valor:4}", end="")
    if col == n - 1:
        print()
```
