```python
# Ejemplos de condiciones con operadores de comparación numéricos
print(5 > 3) # True, porque 5 es mayor que 3
print(5 < 3) # False, porque 5 no es menor que 3
print(5 == 5) # True, porque 5 es igual a 5
print(5 != 3) # True, porque 5 no es igual a 3
print(5 >= 5) # True, porque 5 es mayor o igual a 5
print(5 <= 3) # False, porque 5 no es menor o igual a 3
```

```python
# Tipos de datos que se consideran verdaderos o falsos en Python (booleanos)
print(bool(0)) # False, porque 0 se considera falso en Python
print(bool(1)) # True, porque cualquier número distinto de 0 se considera verdadero en Python
print(bool(-1)) # True, porque cualquier número distinto de 0 se considera verdadero en Python
print(bool("")) # False, porque una cadena vacía se considera falsa en Python
print(bool("hola")) # True, porque una cadena no vacía se considera verdadera en Python
print(bool([])) # False, porque una lista vacía se considera falsa en Python
print(bool([1, 2, 3])) # True, porque una lista no vacía se considera verdadera en Python
print(bool(None)) # False, porque None se considera falso en Python
print(bool(True)) # True, porque True se considera verdadero en Python
print(bool(False)) # False, porque False se considera falso en Python
```

```python
# Juego simple de la lotería, ocultaremos 3 números distintos entre el 0 y el 9
# Pediremos al usuario que introduzca un número de 3 dígitos, 
# y comprobaremos si alguno de esos dígitos coincide con alguno de los números ocultos, devolviendo un mensaje de premio en caso de coincidencia
# premio gordo si coinciden los 3
# o un mensaje de suerte para la próxima en caso contrario
n1, n2, n3 = 1, 2, 3 # números ocultos de la lotería
digito1, digito2, digito3 = input("Introduce tres dígitos separados por espacios: ").split() # la función split() se utiliza para dividir la cadena de entrada en partes, en este caso se espera que el usuario introduzca un número de tres dígitos separados por espacios, y cada dígito se almacena en las variables digito1, digito2 y digito3 respectivamente
digito1 = int(digito1) # se convierte el dígito1 a entero para poder compararlo con los números ocultos
digito2 = int(digito2) # se convierte el dígito2 a entero para poder compararlo con los números ocultos
digito3 = int(digito3) # se convierte el dígito3 a entero para poder compararlo con los números ocultos
if (digito1 == n1 or digito1 == n2 or digito1==n3) and (digito2 == n1 or digito2 == n2 or digito2==n3) and (digito3 == n1 or digito3 == n2 or digito3==n3):
    print("¡Enhorabuena! Has ganado el premio gordo") # imprime un mensaje de premio gordo si los tres dígitos introducidos por el usuario coinciden con los números ocultos de la lotería
else:
    if (digito1 == n1 or digito1 == n2 or digito1==n3) or (digito2 == n1 or digito2 == n2 or digito2==n3) or (digito3 == n1 or digito3 == n2 or digito3==n3):
        print("¡Has tenido suerte! Al menos has acertado un número") # imprime un mensaje de suerte para la próxima si al menos uno de los dígitos introducidos por el usuario coincide con los números ocultos de la lotería
    else:
        print("Lo siento, no has acertado ningún número. ¡Suerte para la próxima!") # imprime un mensaje de suerte para la próxima si ninguno de los dígitos introducidos por el usuario coincide con los números ocultos de la lotería
```
