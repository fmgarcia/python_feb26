```python
cadena = "hola"
if cadena == "hola":
    print("La cadena es igual a 'hola'") # imprime un mensaje si la condición cadena == "hola" es verdadera
```

```python
cadena = "holaaaa"
if cadena == "hola":
    print("La cadena es igual a 'hola'") # imprime un mensaje si la condición cadena == "hola" es verdadera
else:
    print("La cadena no es igual a 'hola'") # imprime un mensaje si la condición cadena == "hola" es falsa
```

```python
anyo = 2024
if anyo % 4 == 0:
    print(f"El año {anyo} es bisiesto") # imprime un mensaje si la condición anyo % 4 == 0 es verdadera, indicando que el año es bisiesto
else:
    print(f"El año {anyo} no es bisiesto") # imprime un mensaje si la condición anyo % 4 == 0 es falsa, indicando que el año no es bisiesto
```

```python
anyo = 2100
if anyo % 4 == 0:
    if anyo % 100 == 0:
        if anyo % 400 == 0:
            print(f"El año {anyo} es bisiesto") # imprime un mensaje si la condición anyo % 400 == 0 es verdadera, indicando que el año es bisiesto
        else:
            print(f"El año {anyo} no es bisiesto") # imprime un mensaje si la condición anyo % 400 == 0 es falsa, indicando que el año no es bisiesto
    else:
        print(f"El año {anyo} es bisiesto") # imprime un mensaje si la condición anyo % 100 == 0 es falsa, indicando que el año es bisiesto    
else:
    print(f"El año {anyo} no es bisiesto") # imprime un mensaje si la condición anyo % 4 == 0 es falsa, indicando que el año no es bisiesto
```

```python
# Ejemplos de condiciones con operadores de comparación numéricos
print(5 > 3) # True, porque 5 es mayor que 3
print(5 < 3) # False, porque 5 no es menor que 3
print(5 == 5) # True, porque 5 es igual a 5
print(5 != 3) # True, porque 5 no es igual a 3
print(5 >= 5) # True, porque 5 es mayor o igual a 5
print(5 <= 3) # False, porque 5 no es menor o igual a 3
```

```python
# Tipos de datos que se consideran verdaderos o falsos en Python (booleanos)
print(bool(0)) # False, porque 0 se considera falso en Python
print(bool(1)) # True, porque cualquier número distinto de 0 se considera verdadero en Python
print(bool(-1)) # True, porque cualquier número distinto de 0 se considera verdadero en Python
print(bool("")) # False, porque una cadena vacía se considera falsa en Python
print(bool("hola")) # True, porque una cadena no vacía se considera verdadera en Python
print(bool([])) # False, porque una lista vacía se considera falsa en Python
print(bool([1, 2, 3])) # True, porque una lista no vacía se considera verdadera en Python
print(bool(None)) # False, porque None se considera falso en Python
print(bool(True)) # True, porque True se considera verdadero en Python
print(bool(False)) # False, porque False se considera falso en Python
```

```python
x = -10
if x:
    print("x es verdadero") # imprime un mensaje si la condición x es verdadera, lo cual es cierto porque cualquier número distinto de 0 se considera verdadero en Python
else:
    print("x es falso") # imprime un mensaje si la condición x es falsa, lo cual no es cierto en este caso porque x es igual a 10, que se considera verdadero en Python
```

```python
nombre = input("Introduce tu nombre: ") # la función input() se utiliza para obtener entrada del usuario, en este caso se le pide al usuario que introduzca su nombre, y ese valor se almacena en la variable "nombre"
if nombre:
    print("Hola, " + nombre + "!") # imprime un mensaje de saludo utilizando el nombre introducido por el usuario, concatenando la variable "nombre" con el mensaje
else:
    print("No has introducido un nombre.") # imprime un mensaje de error si el usuario no ha introducido ningún nombre, lo cual se considera falso en Python porque una cadena vacía se evalúa como falsa
```
