## Operador AND

```python
edad = 17
sexo = "masculino"
if sexo == "masculino" and edad >= 18:
    print("Eres un hombre adulto") # imprime un mensaje si la condición edad >= 18 and sexo == "masculino" es verdadera, lo cual es cierto porque la edad es mayor o igual a 18 y el sexo es masculino
else:
    print("No cumples los requisitos para ser un hombre adulto") # imprime un mensaje si la condición edad >= 18 and sexo == "masculino" es falsa, lo cual no es cierto en este caso porque la condición es verdadera
```

```python
def es_mayor_de_edad(edad):
    print(f"Tu edad es {edad}") # imprime un mensaje que muestra la edad proporcionada por el usuario
    return edad >= 18 # devuelve True si la edad es mayor o igual a 18, lo cual indica que la persona es mayor de edad, y devuelve False en caso contrario

edad = 17
sexo = "femenino"
if sexo == "masculino" and es_mayor_de_edad(edad): # evalua en cortocircuito la condición sexo == "masculino" y es_mayor_de_edad(edad), lo cual es falso porque el sexo es femenino, por lo que no se evalúa la función es_mayor_de_edad(edad) debido al cortocircuito del operador and
    print("Eres un hombre adulto") # imprime un mensaje si la condición edad >= 18 and sexo == "masculino" es verdadera, lo cual es cierto porque la edad es mayor o igual a 18 y el sexo es masculino
else:
    print("No cumples los requisitos para ser un hombre adulto") # imprime un mensaje si la condición edad >= 18 and sexo == "masculino" es falsa, lo cual no es cierto en este caso porque la condición es verdadera
```

## Operador OR

```python
edad = 40
sexo = "femenino"
if sexo == "masculino" or edad >= 18:
    print("Eres un hombre adulto") # imprime un mensaje si la condición edad >= 18 or sexo == "masculino" es verdadera, lo cual es cierto porque el sexo es masculino
else:
    print("No cumples los requisitos para ser un hombre adulto") # imprime un mensaje si la condición edad >= 18 or sexo == "masculino" es falsa, lo cual no es cierto en este caso porque la condición es verdadera
```

## Combinaciones de AND y OR

```python
# Queremos dejar pasar a un gimnasio femenino en el caso que la persona sea mujer y mayor de edad, o sea el hombre dueño del gimnasio, que se llama Fran y tiene 40 años
# Si es Fran además de tener 40 años tiene que tener el dni = "12345678A"
nombre = "Dani"
edad = 40
sexo = "masculino"
dni = "12345678A"
if (sexo == "femenino" and edad >= 18) or (nombre == "Fran" and edad == 40 and dni == "12345678A"):
    print("Puedes entrar al gimnasio") # imprime un mensaje si la condición (sexo == "femenino" and edad >= 18) or (nombre == "Fran" and edad == 40 and dni == "12345678A") es verdadera, lo cual es cierto porque el nombre es Dani y la edad es 40
else:
    print("No puedes entrar al gimnasio") # imprime un mensaje si la condición (sexo == "femenino" and edad >= 18) or (nombre == "Fran" and edad == 40) es falsa, lo cual no es cierto en este caso porque la condición es verdadera
```

## Operador NOT

```python
# Podemos entrar al bingo si no somos menores de edad
menor_de_edad = False
if not menor_de_edad:
    print("Puedes entrar al bingo") # imprime un mensaje si la condición not menor_de_edad es verdadera, lo cual es cierto porque menor_de_edad es False, por lo que not menor_de_edad es True
else:
    print("No puedes entrar al bingo") # imprime un mensaje si la condición not menor_de_edad es falsa, lo cual no es cierto en este caso porque la condición es verdadera
```

```python
"""
Dado un número que el usuario introduce por teclado sacar True o False si es impar
"""
numero = int(input("Introduce un número: ")) # la función input() se utiliza para obtener entrada del usuario, en este caso se le pide al usuario que introduzca un número, y ese valor se convierte a entero con int() y se almacena en la variable "numero"
if not numero % 2 == 0: # evalua si el número es impar, lo cual se determina comprobando si el resto de la división del número entre 2 es diferente de 0
    print(True) # imprime True si la condición numero % 2 != 0 es verdadera, lo cual indica que el número es impar
else:
    print(False) # imprime False si la condición numero % 2 != 0 es falsa, lo cual indica que el número es par
```
