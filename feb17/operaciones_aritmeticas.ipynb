```python
# Factorial de un número (n!) es el producto de todos los números enteros positivos desde 1 hasta n. Por ejemplo, el factorial de 5 (5!) es igual a 1 * 2 * 3 * 4 * 5 = 120. El factorial se utiliza en combinatoria, probabilidad y análisis matemático para contar permutaciones y combinaciones, así como para resolver problemas relacionados con la distribución de objetos.
factorial_5 = 1 * 2 * 3 * 4 * 5
print(factorial_5) # imprime el resultado del factorial de 5, que es 120
```

```python
# IMC = peso / (altura ** 2)
nombre = input("Introduce tu nombre: ")
peso = float(input("Introduce tu peso en kg: "))
altura = float(input("Introduce tu altura en metros: "))
imc = peso / (altura ** 2)
print(f"{nombre}, tu IMC es: {imc:.2f}") # imprime el nombre del usuario y su IMC con dos decimales
```

```python
radio = 4
area_circulo = 3.14159 * (radio ** 2) # calcula el área de un círculo utilizando la fórmula A = π * r^2, donde r es el radio del círculo
print(f"El área del círculo con radio {radio} es: {area_circulo:.2f}") # imprime el área del círculo con dos decimales
circunferencia = 2 * 3.14159 * radio # calcula la circunferencia de un círculo utilizando la fórmula C = 2 * π * r, donde r es el radio del círculo
print(f"La circunferencia del círculo con radio {radio} es: {circunferencia:.2f}") # imprime la circunferencia del círculo con dos decimales
```

```python
# El tipo de datos en Python es dinámico, lo que significa que no es necesario declarar el tipo de una variable antes de asignarle un valor. Python infiere el tipo de datos automáticamente en función del valor asignado a la variable. Por ejemplo, si asignamos un número entero a una variable, Python la tratará como un entero, y si asignamos una cadena de texto, la tratará como una cadena. Esto permite una mayor flexibilidad en la programación, pero también requiere que los programadores sean conscientes de los tipos de datos que están utilizando para evitar errores.
variable = 10 # Python infiere que esta variable es de tipo entero
print(variable) # imprime el valor de la variable, que es 10
variable = "Ahora soy una cadena" # Python infiere que esta variable ahora es de tipo cadena de texto
print(variable) # imprime el nuevo valor de la variable, que es "Ahora soy una cadena
```

```python
numero = 100
numero -= 20 # el operador -= se utiliza para restar el valor de 20 a la variable "numero", lo que es equivalente a escribir numero = numero - 20
print(numero) # imprime el resultado de la resta, que es 80
numero *= 2 # el operador *= se utiliza para multiplicar el valor de "numero" por 2, lo que es equivalente a escribir numero = numero * 2
print(numero) # imprime el resultado de la multiplicación, que es 160
numero /= 4 # el operador /= se utiliza para dividir el valor de "numero" entre 4, lo que es equivalente a escribir numero = numero / 4
print(numero) # imprime el resultado de la división, que es 40.0
numero //= 3 # el operador //= se utiliza para realizar una división entera, lo que es equivalente a escribir numero = numero // 3
print(numero) # imprime el resultado de la división entera, que es 13.0
numero %= 5 # el operador %= se utiliza para obtener el resto de la división de "numero" entre 5, lo que es equivalente a escribir numero = numero % 5
print(numero) # imprime el resultado del operador módulo, que es 3.0
```

```python
"""
Ejercicio 232: Assign the values 30, 60 to the variables width and height, respectively. 
Write a program that uses these two variables to find the area of the rectangle as shown below.
El área del rectángulo es: 1800
"""
width, height = 30, 60 # asignación múltiple, donde width se asigna el valor 30 e height se asigna el valor 60
area_rectangulo = width * height # calcula el área del rectángulo utilizando la fórmula A = width * height
print(f"El área del rectángulo es: {area_rectangulo}") # imprime el área del rectángulo, que es 1800
```

```python
"""
Ejercicio pág 233:
The Pythagorean Theorem states that the square of the hypotenuse c for any right triangle is equal to the square of the base a plus the square of the height b. Write code that calculates the length of the hypotenuse by receiving the base and height as integers. 
"""
a = float(input("Valor de a: "))
b = float(input("Valor de b: "))
c = (a**2 + b**2)**0.5 # calcula la longitud de la hipotenusa utilizando la fórmula c = sqrt(a^2 + b^2), donde a es la base y b es la altura del triángulo rectángulo
print(f"La longitud de la hipotenusa es: {c:.2f}") # imprime la longitud de la hipotenusa con dos decimales
```
