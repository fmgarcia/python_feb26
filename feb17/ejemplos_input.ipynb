```python
x = input("Introduce un número: ") # la función input() se utiliza para obtener entrada del usuario, en este caso se le pide al usuario que introduzca un número, y ese valor se almacena en la variable "x"
if x.isdigit(): # verifica si el valor introducido por el usuario es un número utilizando el método isdigit()
    x = int(x) # convierte la variable "x" a un entero si es un número
    x += 1
    print("El número que has introducido es: " + str(x)) # imprime el número introducido por el usuario, convirtiendo la variable "x" a una cadena de texto para poder concatenarla con el mensaje
else:
    print("Por favor, introduce un número válido.") # imprime un mensaje de error si el valor introducido por el usuario no es un número
```

```python
"""
Ejercicio pág 249:
Write a program that displays the squared values from 2 to 6 in a table as the following. 
As shown below, a can be increased from 2 to 6, and n has a value of 2. 
Input the actual value for the part corresponding to a**n, so that the output result of the equation is 2**2.
a	n	a**n
2	2	4
3	2	9
4	2	16
5	2	25
6	2	36
"""
n = 2
print("a\tn\ta**n") # imprime la cabecera de la tabla con los títulos de las columnas
print(f"2\t{n}\t{2**n}") # imprime la primera fila de la tabla con el valor de a=2, n=2 y a**n=4
print(f"3\t{n}\t{3**n}") # imprime la segunda fila de la tabla con el valor de a=3, n=2 y a**n=9
print(f"4\t{n}\t{4**n}") # imprime la tercera fila de la tabla con el valor de a=4, n=2 y a**n=16
print(f"5\t{n}\t{5**n}") # imprime la cuarta fila de la tabla con el valor de a=5, n=2 y a**n=25
print(f"6\t{n}\t{6**n}") # imprime la quinta fila de la tabla con el valor de a=6, n=2 y a**n=36
```
