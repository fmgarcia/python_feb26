## Diccionarios

### Creación de diccionarios

```python
# Definir un diccionario con claves y valores
juan = { "nombre": "Juan", "edad": 30, "ciudad": "Madrid" }
fran = { "nombre": "Fran", "edad": 25, "ciudad": "Barcelona" }
consuelo = { "nombre": "Consuelo", "edad": 28, "ciudad": "Valencia" }
paco = { "nombre": "Paco", "edad": 35, "ciudad": "Sevilla" }
ana = { "nombre": "Ana", "edad": 22, "ciudad": "Bilbao" }
vacio = {} # crea un diccionario vacío llamado vacio
print(juan) # imprime el diccionario juan, mostrando sus claves y valores
```

#### Acceso a los valores de los campos de un diccionario

```python
# Acceso a los valores de un diccionario utilizando sus claves
print(juan["nombre"]) # imprime el valor asociado a la clave "nombre" en el diccionario juan, mostrando "Juan"
print(fran["edad"]) # imprime el valor asociado a la clave "edad" en eldiccionario fran, mostrando 25
print(consuelo["ciudad"]) # imprime el valor asociado a la clave "ciudad" en el diccionario consuelo, mostrando "Valencia"
```

#### Añadir nuevos campos a un objeto de tipo diccionario

```python
juan["profesion"] = "Ingeniero" # agrega una nueva clave "profesion" con el valor "Ingeniero" al diccionario juan
print(juan) # imprime el diccionario juan después de agregar la nueva clave y valor
```

```python
print(fran["profesion"]) # intenta acceder a la clave "profesion" en el diccionario fran, lo que generará un error KeyError porque esa clave no existe en el diccionario fran
```

#### Recorrer los elementos de un diccionario

```python
for k in juan:
    print(k) # imprime cada clave del diccionario juan en una nueva línea
```

```python
for k,v in juan.items():
    print(f"{k}: {v}") # imprime cada clave y su valor asociado en el diccionario juan en formato "clave: valor"
```

```python
for clave in juan:
    print(f"{clave}: {juan[clave]}") # imprime cada clave del diccionario juan en una nueva línea junto con su valor asociado
```

#### Modificar un elemento del diccionario

```python
juan["edad"] = 31 # actualiza el valor asociado a la clave "edad" en el diccionario juan a 31
print(juan) # imprime el diccionario juan después de actualizar el valor de la clave
```

#### Eliminar un elemento del diccionario

```python
del juan["ciudad"] # elimina la clave "ciudad" y su valor asociado del diccionario juan
print(juan) # imprime el diccionario juan después de eliminar la clave "ciudad
```

#### Métodos de un diccionario

```python
print(juan.keys()) # devuelve una vista de las claves del diccionario juan
print(juan.values()) # devuelve una vista de los valores del diccionario juan
nuevo_juan = juan.fromkeys(["nombre", "edad", "ciudad"], "desconocido") # crea un nuevo diccionario con las claves "nombre", "edad" y "ciudad" y asigna el valor "desconocido" a cada una de ellas
print(nuevo_juan) # imprime el nuevo diccionario creado con fromkeys, mostrando las claves y su valor asociado "desconocido"
```
