#### Crear tuplas

```python
colores = ("rojo", "verde", "azul") # crea una tupla llamada colores que contiene los valores "rojo", "verde" y "azul"
print(colores) # imprime la tupla colores, mostrando los valores que contiene
```

```python
colores[0] = "amarillo" # intenta modificar el primer elemento de la tupla colores, lo cual no es permitido ya que las tuplas son inmutables, lo que resultará en un error de tipo TypeError
```

```python
del colores[0] # intenta eliminar el primer elemento de la tupla colores, lo cual no es permitido ya que las tuplas son inmutables, lo que resultará en un error de tipo TypeError
```

```python
print(colores[0])
```
