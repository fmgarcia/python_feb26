```python
numeros = list(range(2, 11)) # crea una lista llamada numeros que contiene los números del 2 al 10 utilizando la función range() y convirtiéndola en una lista con list()
primos = []
for n in numeros:
    for i in range(2, n):
        if n % i == 0:
            break
    else: # el bloque else se ejecuta si el bucle for no se rompe, lo que significa que n es un número primo
        primos.append(n)
print(primos) # imprime la lista de números primos encontrados en la lista numeros utilizando un bucle for para iterar sobre cada número y verificar si es primo, agregándolo a la lista primos si lo es
```

```python
numeros = list(range(2, 11)) # crea una lista llamada numeros que contiene los números del 2 al 10 utilizando la función range() y convirtiéndola en una lista con list()
primos = []
for n in numeros:
    primo = True
    for i in range(2, n):
        if n % i == 0:
            primo = False
    if primo: # verifica si la variable primo es True, lo que indicaría que n es un número primo
        primos.append(n) # si n es un número primo, lo agrega a la lista primos
print(primos) # imprime la lista de números primos encontrados en la lista numeros utilizando un bucle for para iterar sobre cada número y verificar si es primo, agregándolo a la lista primos si lo es
```

```python
numeros = list(range(2, 11)) # crea una lista llamada numeros que contiene los números del 2 al 10 utilizando la función range() y convirtiéndola en una lista con list()
primos = []
for n in numeros:
    primo = True # asume que n es un número primo al inicio de cada iteración del bucle for
    i = 2
    while i < n and primo: # mientras i sea menor que n, lo que indica que aún no se ha verificado si n es divisible por algún número entre 2 y n-1
        if n % i == 0:
            primo = False # si n es divisible por algún número entre 2 y n-1, entonces no es un número primo, por lo que se establece la variable primo en False para salir del bucle while
        i += 1 # incrementa n para verificar el siguiente número en la próxima iteración del bucle while
    if primo: # si la variable primo sigue siendo True después de verificar todos los posibles divisores, entonces n es un número primo
        primos.append(n) # si n es un número primo, lo agrega a la lista primos
print(primos) # imprime la lista de números primos encontrados en la lista numeros utilizando un bucle for para iterar sobre cada número y verificar si es primo, agregándolo a la lista primos si lo es
```

```python
# Ejercicio pag 86 módulo 2
s_list = ['abc', 'bcd', 'bcdefg', 'abba', 'cddc', 'op']
mas_pequeno = s_list[0]
for s in s_list[1:]:
    if len(s) < len(mas_pequeno):
        mas_pequeno = s
print("El string más pequeño es:", mas_pequeno)
```

```python
s_list = ['abc', 'bcd', 'bcdefg', 'abba', 'cddc', 'opq']
sorted_list = sorted(s_list, key=len)
longitud_mas_corta = len(sorted_list[0])
mas_cortas = []
for s in sorted_list:
    if len(s) == longitud_mas_corta:
        mas_cortas.append(s)
    else:
        break
print(f"Las cadenas con menos letras son: {mas_cortas}")
```
