```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.plot([1, 2, 3], [4, 5, 6])
plt.show()
```

```python
x = range(100)
y = [xi**2 for xi in x]
plt.plot(x, y, linewidth=5.0, color='red')
plt.title("Gráfico de x vs x^2")
plt.xlabel("x")
plt.ylabel("x^2")
plt.show()
```

```python
x = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo']
y = [100, 150, 120, 250, 200]
plt.bar(x, y, color='blue')
plt.title("Ventas mensuales")
plt.xlabel("Meses")
plt.ylabel("Ventas")
plt.show()
```

```python
plt.figure(figsize=(13, 6))
plt.plot(x, y, marker='o', linestyle='--', color='green')
plt.title("Gráfico de ventas mensuales")
plt.xlabel("Meses")
plt.ylabel("Ventas")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(13, 6))
plt.xticks(rotation=45, size=12)
plt.yticks(rotation=45, size=12)
plt.barh(x, y, color='orange')
plt.title("Gráfico de ventas mensuales")
plt.xlabel("Meses")
plt.ylabel("Ventas")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.pie(y, labels=x, autopct='%1.1f%%', startangle=180)
plt.title("Distribución de ventas")
plt.show()
```
