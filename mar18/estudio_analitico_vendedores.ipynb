```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
vendedores = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
meses = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio', 'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre']
datos = np.random.randint(50, 200, size=(len(vendedores), len(meses)))
```

```python
datos
```

```python
# Grafico de líneas para cada vendedor
plt.figure(figsize=(12, 6))
for i, vendedor in enumerate(vendedores):
    plt.plot(meses, datos[i], marker='o', label=vendedor)
plt.title("Evolución de ventas por vendedor")
plt.xlabel("Meses")
plt.ylabel("Ventas")
plt.legend()
plt.show()
```

```python
# Obtener gráfico a nivel de cada vendedor
for i, vendedor in enumerate(vendedores):
    plt.figure(figsize=(12, 6))
    plt.bar(meses, datos[i], label=vendedor, color=np.random.rand(3,))
    plt.title("Ventas mensuales por vendedor")
    plt.xlabel("Meses")
    plt.ylabel("Ventas")
    plt.legend()
    plt.show()
```
