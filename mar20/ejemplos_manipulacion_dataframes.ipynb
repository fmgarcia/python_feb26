```python
republicas_europeas = df_paises[(df_paises['Forma de gobierno'] == 'República parlamentaria') & (df_paises['Continente'] == 'Europa')] # Crea un nuevo DataFrame llamado republicas_europeas que contiene solo las filas del DataFrame df_paises donde la columna 'Tipo de gobierno' es igual a 'República parlamentaria' y la columna 'Continente' es igual a 'Europa'.
republicas_europeas.reset_index(drop=True, inplace=True) # Restablece los índices del DataFrame republicas_europeas, eliminando los índices anteriores y creando nuevos índices secuenciales. El parámetro drop=True evita que los índices anteriores se agreguen como una nueva columna en el DataFrame.
republicas_europeas.head()
```

```python
republicas_europeas_africanas = df_paises[(df_paises['Forma de gobierno'] == 'República parlamentaria') & (df_paises['Continente'].isin(['Europa', 'África']))] # Crea un nuevo DataFrame llamado republicas_europeas_africanas que contiene solo las filas del DataFrame df_paises donde la columna 'Forma de gobierno' es igual a 'República parlamentaria' y la columna 'Continente' está en la lista ['Europa', 'África'].
republicas_europeas_africanas.reset_index(drop=True, inplace=True) # Restablece los índices del DataFrame republicas_europeas_africanas, eliminando los índices anteriores y creando nuevos índices secuenciales. El parámetro drop=True evita que los índices anteriores se agreguen como una nueva columna en el DataFrame.
republicas_europeas_africanas.head()
```

Renombrar columnas

```python
df_paises = df_paises.rename(columns={'Estado soberano (nombre común)': 'Pais','Estado soberano (nombre común).1': 'Pais_2','Forma de gobierno': 'Gobierno'}) # Cambia el nombre de las columnas 'Estado soberano (nombre común)' a 'Pais' y 'Forma de gobierno' a 'Gobierno' en el DataFrame df_paises.
df_paises.head()
```

```python
df_personas = pd.DataFrame({'Nombre': ['Alice', 'Bob', 'Charlie', 'Fran', 'Dani'], 'Edad': [25, 30, np.nan, 40, 45], 'Ciudad': ['Madrid', np.nan, 'Valencia', 'Sevilla', 'Bilbao'], 'Sexo': ['Femenino', 'Masculino', 'Masculino', np.nan, 'Masculino'], 'Altura': [165, np.nan, 175, 170, 185], 'Peso': [55, 70, np.nan, 60, 75]}) # Crea un nuevo DataFrame llamado df_personas con cinco columnas: 'Nombre', 'Edad', 'Ciudad', 'Sexo' y 'Altura', y tres filas de datos correspondientes a cada persona.
df_personas.head()
```

```python
df_personas.describe()
```

```python
df_personas['Alto_booleano'] = df_personas['Altura'] >= 180 # Crea una nueva columna llamada 'Alto' en el DataFrame df_personas que contiene valores booleanos (True o False) dependiendo de si la altura de cada persona es mayor a 180 cm.
df_personas['Alto_texto'] = np.where(df_personas['Altura'] >= 180, 'Sí', 'No') # Crea una nueva columna llamada 'Alto_texto' en el DataFrame df_personas que contiene valores de texto ('Sí' o 'No') dependiendo de si la altura de cada persona es mayor a 180 cm.
df_personas.head()
```

Agrupaciones y agregaciones

```python
pesos_por_sexo = df_personas.groupby('Sexo')['Peso'].sum().reset_index() # Agrupa el DataFrame df_personas por la columna 'Sexo', calcula la suma de la columna 'Peso' para cada grupo y restablece los índices del resultado para obtener un nuevo DataFrame llamado pesos_por_sexo.
pesos_por_sexo
```

```python
estadisticas_por_sexo = df_personas.groupby('Sexo')['Peso'].agg(Media='mean', Suma='sum', Cuenta='count').reset_index() # Agrupa el DataFrame df_personas por la columna 'Sexo', calcula la media, la suma y el conteo de la columna 'Peso' para cada grupo y restablece los índices del resultado para obtener un nuevo DataFrame llamado estadisticas_por_sexo.
estadisticas_por_sexo
```

Ordenaciones

```python
df_personas_ordenado = df_personas.sort_values(by='Edad', ascending=False).reset_index(drop=True) # Ordena el DataFrame df_personas por la columna 'Edad' en orden descendente y guarda el resultado en un nuevo DataFrame llamado df_personas_ordenado.
df_personas_ordenado.head()
```

```python
df_ordenado_multiple = df_personas.sort_values(by=['Nombre', 'Ciudad'], ascending=[True, False]).reset_index(drop=True) # Ordena el DataFrame df_personas por la columna 'Sexo' en orden ascendente y luego por la columna 'Edad' en orden descendente, y guarda el resultado en un nuevo DataFrame llamado df_ordenado_multiple.
df_ordenado_multiple.head()
```
