### Creación de Dataframes

A partir de un Diccionario

```python
import pandas as pd
```

```python
data = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Marta'],
    'Edad': [28, 34, 22, 45],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}
df = pd.DataFrame(data)
print(df)
```

A partir de una lista

```python
data_lista = [
    ['Ana', 28, 'Madrid'],
    ['Luis', 34, 'Barcelona'],
    ['Carlos', 22, 'Valencia'],
    ['Marta', 45, 'Sevilla']
]

df_lista = pd.DataFrame(data_lista, columns=['Nombre', 'Edad', 'Ciudad'])
print(df_lista)
```

A partir de una lista de diccionarios (Ideal para APIs/JSON)

```python
data_json = [
    {"Nombre": "Ana", "Edad": 28, "Ciudad": "Madrid"},
    {"Nombre": "Luis", "Edad": 34, "Ciudad": "Barcelona"},
    {"Nombre": "Carlos", "Edad": 22, "Ciudad": "Valencia"},
    {"Nombre": "Marta", "Edad": 45, "Ciudad": "Sevilla"}
]
df_json = pd.DataFrame(data_json)
print(df_json)
```

A partir de un diccionario de diccionarios

```python
data_dict_dict = {
    'Nombre': {0: 'Ana', 1: 'Luis', 2: 'Carlos', 3: 'Marta'},
    'Edad': {0: 28, 1: 34, 2: 22, 3: 45},
    'Ciudad': {0: 'Madrid', 1: 'Barcelona', 2: 'Valencia', 3: 'Sevilla'}
}
df_dict_dict = pd.DataFrame(data_dict_dict)
print(df_dict_dict)
```

DataFrame vacío con columnas predefinidas

```python
df_vacio = pd.DataFrame(columns=['Nombre', 'Edad', 'Ciudad'])
print(df_vacio)
```

Dataframe a partir de Series de Pandas

```python
series_nombre = pd.Series(['Ana', 'Luis', 'Carlos', 'Marta'])
series_edad = pd.Series([28, 34, 22, 45])
series_ciudad = pd.Series(['Madrid', 'Barcelona', 'Valencia', 'Sevilla'])
df_series = pd.DataFrame({
    'Nombre': series_nombre,
    'Edad': series_edad,
    'Ciudad': series_ciudad
})
print(df_series)
```

```python
series_nombre = pd.Series(['Ana', 'Luis', 'Carlos', 'Marta'])
series_edad = pd.Series([28, 34, 22, 45])
series_ciudad = pd.Series(['Madrid', 'Barcelona', 'Valencia', 'Sevilla'])
df_series = pd.concat([series_nombre, series_edad, series_ciudad], axis=1) # Concatenar las series como columnas (eso lo indica axis=1)
df_series.columns = ['Nombre', 'Edad', 'Ciudad']
print(df_series)
```

A partir de datos ficticios

```python
num_filas = 10000
df_grande = pd.DataFrame({
    'Nombre': ['Nombre' + str(i) for i in range(num_filas)],
    'Edad': [i % 100 for i in range(num_filas)],
    'Ciudad': ['Ciudad' + str(i % 10) for i in range(num_filas)]
}, index=range(1, num_filas + 1))
print(df_grande)
```

```python
fila_base = pd.DataFrame([{
    'Nombre': 'NombreX',
    'Edad': 25,
    'Ciudad': 'CiudadX'
}])
# Repetir la fila base 10000 veces para crear un DataFrame grande
df_grande_repetido = pd.concat([fila_base] * num_filas, ignore_index=True)
print(df_grande_repetido)
```

A partir de un csv

```python
# Lo más común en un CSV es que la primera fila contenga los nombres de las columnas, por lo que no es necesario especificar el parámetro header. Sin embargo, si el CSV no tiene una fila de encabezado, puedes usar header=None para indicar que la primera fila es parte de los datos.
# Los separadores comunes en un CSV son la coma (,) y el punto y coma (;). Si tu CSV utiliza un separador diferente, asegúrate de especificarlo correctamente con el parámetro sep.
data_csv = pd.read_csv('../data/datos_ficticios.csv', sep=',', encoding='utf-8')
data_csv.head()
```
