### Creación de Dataframes

A partir de un Diccionario

```python
import pandas as pd
```

```python
data = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Marta'],
    'Edad': [28, 34, 22, 45],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}
df = pd.DataFrame(data)
print(df)
```

A partir de una lista

```python
data_lista = [
    ['Ana', 28, 'Madrid'],
    ['Luis', 34, 'Barcelona'],
    ['Carlos', 22, 'Valencia'],
    ['Marta', 45, 'Sevilla']
]

df_lista = pd.DataFrame(data_lista, columns=['Nombre', 'Edad', 'Ciudad'])
print(df_lista)
```

Dataframe a partir de Series de Pandas

```python
series_nombre = pd.Series(['Ana', 'Luis', 'Carlos', 'Marta'])
series_edad = pd.Series([28, 34, 22, 45])
series_ciudad = pd.Series(['Madrid', 'Barcelona', 'Valencia', 'Sevilla'])
df_series = pd.DataFrame({
    'Nombre': series_nombre,
    'Edad': series_edad,
    'Ciudad': series_ciudad
})
print(df_series)
```

```python
series_nombre = pd.Series(['Ana', 'Luis', 'Carlos', 'Marta'])
series_edad = pd.Series([28, 34, 22, 45])
series_ciudad = pd.Series(['Madrid', 'Barcelona', 'Valencia', 'Sevilla'])
df_series = pd.concat([series_nombre, series_edad, series_ciudad], axis=1) # Concatenar las series como columnas (eso lo indica axis=1)
df_series.columns = ['Nombre', 'Edad', 'Ciudad']
print(df_series)
```

A partir de datos ficticios

```python
num_filas = 10000
df_grande = pd.DataFrame({
    'Nombre': ['Nombre' + str(i) for i in range(num_filas)],
    'Edad': [i % 100 for i in range(num_filas)],
    'Ciudad': ['Ciudad' + str(i % 10) for i in range(num_filas)]
}, index=range(1, num_filas + 1))
print(df_grande)
```

```python
fila_base = pd.DataFrame([{
    'Nombre': 'NombreX',
    'Edad': 25,
    'Ciudad': 'CiudadX'
}])
# Repetir la fila base 10000 veces para crear un DataFrame grande
df_grande_repetido = pd.concat([fila_base] * num_filas, ignore_index=True)
print(df_grande_repetido)
```

A partir de un html

```python
import pandas as pd
import requests
from io import StringIO

url_base = 'https://es.wikipedia.org/wiki/Anexo:Países'

# 1. Definimos un User-Agent simulando ser el navegador Google Chrome en Windows
cabeceras = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

# 2. Hacemos la petición incluyendo las cabeceras
respuesta = requests.get(url_base, headers=cabeceras)

# 3. Verificamos que Wikipedia nos ha dado acceso (el código 200 significa "OK")
if respuesta.status_code == 200:
    # Le pasamos el texto HTML a pandas usando StringIO
    df_wikipedia = pd.read_html(StringIO(respuesta.text))
    
    # Extraemos nuestra tabla. A veces en Wikipedia la tabla principal no es la 0,
    # pero en este artículo específico suele ser la primera [0] o la segunda [1].
    paises = df_wikipedia[0]
    
    
else:
    print(f"Error al acceder a la página. Código: {respuesta.status_code}")
paises.head()
```
