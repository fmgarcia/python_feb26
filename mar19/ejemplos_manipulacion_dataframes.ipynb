```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Primero cargo el dataset en un DataFrame

```python
df = pd.read_csv('../data/paises.csv', header="infer", encoding="utf-8")
df.head()
```

```python
df.info()
```

Eliminar una columna

```python
df.drop(columns=['Ubicación']).head() # Elimina la columna 'Ubicación' pero NO MODIFICA el DataFrame original y muestra las primeras filas del DataFrame resultante.
```

```python
df.head()
```

```python
df_sin_ubicacion = df.drop(columns=['Ubicación']) # Elimina la columna 'Ubicación' y guarda el resultado en un nuevo DataFrame llamado df_sin_ubicacion.
df_sin_ubicacion.head()
```

Eliminación de columnas sobre el propio DataFrame

```python
df.drop(columns=['Ubicación'], inplace=True) # Elimina la columna 'Ubicación' del DataFrame original df y modifica el DataFrame original.
df.head()
```

```python
# Eliminar filas impares del Dataframe
df.drop(index=df.index[1::2], axis=0, inplace=True) # Elimina las filas con índices impares del DataFrame df_sin_ubicacion y modifica el DataFrame original.
df.head()
```

```python
df.to_csv('../data/paises_limpios.csv', index=False) # Guarda el DataFrame df_sin_ubicacion en un nuevo archivo CSV llamado 'paises_limpios.csv' sin incluir los índices de las filas.
```

Seleccionar filas

```python
df_indices = df.iloc[0:10] # Crea un nuevo DataFrame llamado df_indices que contiene las primeras 10 filas del DataFrame df utilizando la función iloc para seleccionar las filas por su posición.
df_indices
```
