## Diccionarios

### Creación de diccionarios

```python
# Definir un diccionario con claves y valores
juan = { "nombre": "Juan", "edad": 30, "ciudad": "Madrid" }
fran = { "nombre": "Fran", "edad": 25, "ciudad": "Barcelona" }
consuelo = { "nombre": "Consuelo", "edad": 28, "ciudad": "Valencia" }
paco = { "nombre": "Paco", "edad": 35, "ciudad": "Sevilla" }
ana = { "nombre": "Ana", "edad": 22, "ciudad": "Bilbao" }
vacio = {} # crea un diccionario vacío llamado vacio
print(juan) # imprime el diccionario juan, mostrando sus claves y valores
```

#### Acceso a los valores de los campos de un diccionario

```python
# Acceso a los valores de un diccionario utilizando sus claves
print(juan["nombre"]) # imprime el valor asociado a la clave "nombre" en el diccionario juan, mostrando "Juan"
print(fran["edad"]) # imprime el valor asociado a la clave "edad" en eldiccionario fran, mostrando 25
print(consuelo["ciudad"]) # imprime el valor asociado a la clave "ciudad" en el diccionario consuelo, mostrando "Valencia"
```

#### Añadir nuevos campos a un objeto de tipo diccionario

```python
juan["profesion"] = "Ingeniero" # agrega una nueva clave "profesion" con el valor "Ingeniero" al diccionario juan
print(juan) # imprime el diccionario juan después de agregar la nueva clave y valor
```

```python
print(fran["profesion"]) # intenta acceder a la clave "profesion" en el diccionario fran, lo que generará un error KeyError porque esa clave no existe en el diccionario fran
```

#### Recorrer los elementos de un diccionario

```python
for k in juan:
    print(k) # imprime cada clave del diccionario juan en una nueva línea
```

```python
for k,v in juan.items():
    print(f"{k}: {v}") # imprime cada clave y su valor asociado en el diccionario juan en formato "clave: valor"
```

```python
for clave in juan:
    print(f"{clave}: {juan[clave]}") # imprime cada clave del diccionario juan en una nueva línea junto con su valor asociado
```

#### Modificar un elemento del diccionario

```python
juan["edad"] = 31 # actualiza el valor asociado a la clave "edad" en el diccionario juan a 31
print(juan) # imprime el diccionario juan después de actualizar el valor de la clave
```

#### Eliminar un elemento del diccionario

```python
del juan["ciudad"] # elimina la clave "ciudad" y su valor asociado del diccionario juan
print(juan) # imprime el diccionario juan después de eliminar la clave "ciudad
```

#### Métodos de un diccionario

```python
print(juan.keys()) # devuelve una vista de las claves del diccionario juan
print(juan.values()) # devuelve una vista de los valores del diccionario juan
nuevo_juan = juan.fromkeys(["nombre", "edad", "ciudad"], "desconocido") # crea un nuevo diccionario con las claves "nombre", "edad" y "ciudad" y asigna el valor "desconocido" a cada una de ellas
print(nuevo_juan) # imprime el nuevo diccionario creado con fromkeys, mostrando las claves y su valor asociado "desconocido"
```

#### Funciones del diccionario

```python
persona = dict(nombre="Laura", edad=27, ciudad="Granada") # crea un nuevo diccionario utilizando la función dict con claves y valores especificados
print(len(persona)) # devuelve el número de pares clave-valor en el diccionario persona, que es 3
print("nombre" in persona) # verifica si la clave "nombre" existe en el diccionario persona, lo que devuelve True
print("profesion" in persona) # verifica si la clave "profesion" existe en el diccionario persona, lo que devuelve False
print("Laura" in persona) # verifica si el valor "Laura" existe en el diccionario persona, lo que devuelve False porque se busca como clave, no como valor
print("Laura" in persona.keys()) # verifica si el valor "Laura" existe entre las claves del diccionario persona, lo que devuelve False
print("Granada" in persona.values()) # verifica si el valor "Granada" existe entre los valores del diccionario persona, lo que devuelve True
print("Granada" not in persona.values()) # verifica si el valor "Granada" existe entre los valores del diccionario persona, lo que devuelve True
```

```python
persona1 = {"nombre": "Laura", "edad": 27, "ciudad": "Granada"} # crea un nuevo diccionario persona1 con claves y valores especificados
persona2 = {"nombre": "Laura", "ciudad": "Granada", "edad": 27} # crea un nuevo diccionario persona2 con las mismas claves y valores que persona1
print(persona1 == persona2) # compara los diccionarios persona1 y persona2, lo que devuelve True porque tienen las mismas claves y valores, aunque el orden de las claves es diferente
```

```python
if (persona1 > persona2):
    print("persona1 es mayor que persona2")
```

#### Json como diccionario

```python
compra = { "name"   : "John Smith",
  "sku"    : "20223",
  "price"  : 23.95,
  "shipTo" : { "name" : "Jane Smith",
               "address" : "123 Maple Street",
               "city" : "Pretendville",
               "state" : "NY",
               "zip"   : "12345" },
  "billTo" : { "name" : "John Smith",
               "address" : "123 Maple Street",
               "city" : "Pretendville",
               "state" : "NY",
               "zip"   : "12345" }
}

print(compra["name"]) # John Smith
print(compra["price"]) # 23.95
print(compra["shipTo"]["name"]) # Jane Smith
print(compra["billTo"]["address"]) # 123 Maple Street
```

```python
import requests

url = "https://swapi.info/api/people/1"
response = requests.get(url) # realiza una solicitud GET a la URL especificada y almacena la respuesta en la variable response
data = response.json() # convierte la respuesta en formato JSON a un diccionario de Python y lo almacena en la variable data
print(data["name"]) # Luke Skywalker
print(data["height"]) # 172
```

```python
import requests

url = "https://swapi.info/api/people/"

for i in range(1, 11):
    url_persona = url + str(i) # construye la URL para cada persona concatenando la URL base con el número de persona
    response = requests.get(url_persona) # realiza una solicitud GET a la URL de cada persona y almacena la respuesta en la variable response
    data = response.json() # convierte la respuesta en formato JSON a un diccionario de Python
    print(f"La persona {i} es: {data['name']}") # imprime el nombre de cada persona obtenida de la API
    
```

```python
personas = []
personas.append({"name": "Luke Skywalker", "height": "172"})
personas.append({"name": "C-3PO", "height": "167"})
print(personas) # imprime la lista de diccionarios personas, mostrando cada diccionario con su clave "name" y "height"
```

```python
import requests

url = "https://swapi.info/api/people/"
personas = [] # Creo una lista inicialmente vacía para almacenar los diccionarios de cada persona obtenida de la API

for i in range(1, 11):
    url_persona = url + str(i) # construye la URL para cada persona concatenando la URL base con el número de persona
    response = requests.get(url_persona) # realiza una solicitud GET a la URL de cada persona y almacena la respuesta en la variable response
    data = response.json() # convierte la respuesta en formato JSON a un diccionario de Python
    print("Leyendo datos de la persona", i) # imprime un mensaje indicando que se están leyendo los datos de la persona actual
    personas.append(data) # agrega el diccionario de cada persona a la lista personas

print("Carga de datos finalizada")
print(f"Se cargaron {len(personas)} personas")
```

```python
print(personas) # imprime la lista completa de diccionarios personas, mostrando los datos de cada persona obtenida de la API
```

```python
import requests

url_base = "https://api.imdbapi.dev/titles/"
peliculas = []
```

```python
 # Creo una lista inicialmente vacía para almacenar los diccionarios de cada persona obtenida de la API

id_pelicula = input("Introduce el ID de la película (por ejemplo, tt0111161 para 'The Shawshank Redemption'): ") # Solicita al usuario que introduzca el ID de la película
url_pelicula = url_base + id_pelicula # Construye la URL completa para la película específica
response = requests.get(url_pelicula) # realiza una solicitud GET a la URL de la película y almacena la respuesta en la variable response
data = response.json() # convierte la respuesta en formato JSON a un diccionario de Python
peliculas.append(data) # agrega el diccionario de la película a la lista peliculas

```

```python
print(peliculas) # imprime la lista completa de diccionarios peliculas, mostrando los datos de cada película obtenida de la API
```

```python
for pelicula in peliculas:
    print(f"Título: {pelicula['primaryTitle']}, Año: {pelicula['startYear']}") # imprime el título y el año de cada película en la lista peliculas
```
