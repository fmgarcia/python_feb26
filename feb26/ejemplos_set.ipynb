### Crear set

```python
miset = set() # set vacío
miset2 = {1, 2, 3, 4, 5} # set con elementos
lista = [1, 2, 3, 4, 5, 5, 5, 5] # lista con elementos repetidos
miset3 = set(lista) # set creado a partir de una lista, elimina los duplicados
print(miset) # set vacío
print(miset2) # set con elementos
print(miset3) # set creado a partir de una lista, sin duplicados
```

#### Añadir elementos a un set

```python
miset.add(6) # añadimos un elemento al set
print(miset) # set con el nuevo elemento
```

#### Borrar elementos de un set

```python
miset.remove(6) # eliminamos un elemento del set
print(miset) # set sin el elemento eliminado
```

#### Operaciones de conjuntos con set

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
print(set1.union(set2)) # unión de dos sets
print(set1.intersection(set2)) # intersección de dos sets
print(set1.difference(set2)) # diferencia de dos sets
print(set2.difference(set1)) # diferencia de dos sets
print(set1.symmetric_difference(set2)) # diferencia simétrica de dos sets
```

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
print(set1 | set2) # unión de dos sets
print(set1 & set2) # intersección de dos sets
print(set1 - set2) # diferencia de dos sets
print(set2 - set1) # diferencia de dos sets
print(set1 ^ set2) # diferencia simétrica de dos sets
```
