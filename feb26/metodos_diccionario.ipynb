### Métodos

#### Setdefault

```python
diccionario = {}
nombres = ["Fran", "Ana", "Luis"]
for nombre in nombres:
    diccionario[nombre] = diccionario.setdefault(nombre, 0)
print(diccionario)
```

```python
diccionario.setdefault("Dani", 0)
print(diccionario)
```

```python
diccionario["Paco"] = 0
print(diccionario)
```

#### update

```python
diccionario.update({"Paco": 1})
print(diccionario)
```

```python
diccionario["Paco"] = 2
print(diccionario)
```

```python
diccionario.update({"Francisco": 1})
print(diccionario)
```

#### pop

```python
eliminado = diccionario.pop("Paco")
print(f"He eliminado a Paco, que tenía el valor {eliminado}")
print(diccionario)
```

#### clear

```python
diccionario.clear()
print(diccionario)
```

#### get

```python
diccionario = {'Fran': 1, 'Ana': 0, 'Luis': 0, 'Dani': 0, 'Paco': 0}
print(diccionario.get("Fran")) # diccionario["Fran"] también funciona, pero get es más seguro
```

```python
print(diccionario["Francisco"]) # Esto da error, porque Francisco no existe en el diccionario
```

```python
print(diccionario.get("Francisco",0)) # Esto no da error, porque get devuelve el segundo argumento si la clave no existe
```

#### keys, values, items

```python
diccionario.keys() # Devuelve las claves del diccionario
```

```python
diccionario.values() # Devuelve los valores del diccionario
```

```python
diccionario
```

```python
diccionario.items() # Devuelve una lista de tuplas (clave, valor) del diccionario
```

```python
datos = list(diccionario.items())
print("La segunda persona es", datos[1][0], "y su valor es", datos[1][1])
```

```python
person_dic = {'Last Name': 'Doe', 'First Name': 'David', 'Company': 'Samsung'}
for key, value in person_dic.items():
    print(f"{key}: {value}")
```

```python
person_dic = {'Last Name': 'Doe', 'First Name': 'David', 'Company': 'Samsung'}
for key in person_dic.keys():
    print(f"{key}: {person_dic[key]}")
```

```python
person_dic = {'Last Name': 'Doe', 'First Name': 'David', 'Company': 'Samsung'}
for key in person_dic.keys():
    print(f"{key}: {person_dic.get(key)}")
```
