### Métodos

#### Setdefault

```python
diccionario = {}
nombres = ["Fran", "Ana", "Luis"]
for nombre in nombres:
    diccionario[nombre] = diccionario.setdefault(nombre, 0)
print(diccionario)
```

```python
diccionario.setdefault("Dani", 0)
print(diccionario)
```

```python
diccionario["Paco"] = 0
print(diccionario)
```

#### update

```python
diccionario.update({"Paco": 1})
print(diccionario)
```

```python
diccionario["Paco"] = 2
print(diccionario)
```

```python
diccionario.update({"Francisco": 1})
print(diccionario)
```

#### pop

```python
eliminado = diccionario.pop("Paco")
print(f"He eliminado a Paco, que tenía el valor {eliminado}")
print(diccionario)
```

#### clear

```python
diccionario.clear()
print(diccionario)
```

#### get

```python
diccionario = {'Fran': 1, 'Ana': 0, 'Luis': 0, 'Dani': 0, 'Paco': 0}
print(diccionario.get("Fran")) # diccionario["Fran"] también funciona, pero get es más seguro
```

```python
print(diccionario["Francisco"]) # Esto da error, porque Francisco no existe en el diccionario
```

```python
print(diccionario.get("Francisco",0)) # Esto no da error, porque get devuelve el segundo argumento si la clave no existe
```

#### fromkeys

```python
# Crear un diccionario a partir de una lista
nombres = ["David", "Ana", "Luis", "María"]
diccionario = dict.fromkeys(nombres, 0) # Crea un diccionario con las claves de la lista y el valor 0
print(diccionario)
```

#### Copiar diccionarios

```python
nombres = {"David": 10, "Ana": 20, "Luis": 30, "María": 40}
nombres2 = nombres # Esto no crea una copia del diccionario, sino que ambas variables apuntan al mismo diccionario
nombres2["David"] = 100 # Esto cambia el valor de "David" en ambos diccionarios, porque son el mismo diccionario
print(nombres)
```

```python
nombres = {"David": 10, "Ana": 20, "Luis": 30, "María": 40}
nombres2 = nombres.copy() # Crea una copia del diccionario
nombres2["David"] = 100 # Esto cambia el valor de "David" solo en el diccionario nombres2, porque son diccionarios diferentes
print(nombres)
```

```python
# copy es una copia superficial del diccionario, lo que significa que si el valor de una clave es un diccionario, ambos diccionarios apuntan al mismo diccionario, por lo que si se modifica el diccionario interno, se modifica en ambos diccionarios
nombres = {"David": {"Last Name": "Doe", "Company": "Samsung"},
        "Ana": {"Last Name": "Smith", "Company": "Google"},
        "Luis": {"Last Name": "Garcia", "Company": "Microsoft"},
        "María": {"Last Name": "Lopez", "Company": "Apple"}
}
nombres2 = nombres.copy() # Crea una copia del diccionario, pero los valores siguen siendo los mismos diccionarios
nombres2["David"]["Last Name"] = "Smith" # Esto cambia el valor de "Last Name" de "David" en el diccionario nombres2, pero no en nombres, porque nombres2 es una copia superficial del diccionario nombres
print(nombres["David"]["Last Name"]) # Imprime "Doe", porque el diccionario original no fue modificado
print(nombres2["David"]["Last Name"]) # Imprime "Smith", porque el diccionario copiado fue modificado
```

```python
# deepcopy es una función del módulo copy que crea una copia profunda de un objeto, es decir, también copia los objetos internos, en lugar de solo copiar las referencias a ellos
import copy
nombres = {"David": {"Last Name": "Doe", "Company": "Samsung"},
        "Ana": {"Last Name": "Smith", "Company": "Google"},
        "Luis": {"Last Name": "Garcia", "Company": "Microsoft"},
        "María": {"Last Name": "Lopez", "Company": "Apple"}
}
nombres2 = copy.deepcopy(nombres) # Crea una copia profunda del diccionario, es decir, también copia los diccionarios internos
nombres2["David"]["Last Name"] = "Smith" # Esto cambia el valor de  "Last Name" de "David" en el diccionario nombres2, pero no en nombres, porque nombres2 es una copia profunda del diccionario nombres
print(nombres["David"]["Last Name"]) # Imprime "Doe", porque el diccionario original no fue modificado
print(nombres2["David"]["Last Name"]) # Imprime "Smith", porque el diccionario copiado fue modificado
```
