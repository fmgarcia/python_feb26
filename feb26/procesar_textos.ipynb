#### Separar palabras

```python
cadena = "Hola, esto es una cadena de texto. con comas"
division_basica = cadena.split(",") # Devuelve una lista de palabras, separadas por comas
print(division_basica)
division_palabras = cadena.split() # Devuelve una lista de palabras, separadas por espacios
print(division_palabras)
division_avanzada = [palabra.strip() for palabra in cadena.split(",")] # Devuelve una lista de palabras, separadas por comas y sin espacios
print(division_avanzada)
division_final = cadena.replace(",","").replace(".","").split() # Devuelve una lista de palabras, separadas por espacios y sin comas ni puntos
print(division_final)
```

#### Join

```python
nombres = ["David", "Ana", "Luis", "María"]
cadena_nombres = ", ".join(nombres) # Devuelve una cadena de texto con los nombres separados por comas
print(cadena_nombres)
nombres_separados_punto_coma = ";".join(nombres) # Devuelve una cadena de texto con los nombres separados por punto y coma 
print(nombres_separados_punto_coma)
```
