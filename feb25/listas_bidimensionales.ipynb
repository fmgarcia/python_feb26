### Declaración

```python
fila1 = [0,1,0,0,1]
fila2 = [1,0,1,0,0]
fila3 = [0,1,0,1,0]

cine = [fila1, fila2, fila3] # lista de listas. Almacena la información de las filas del cine. Cada fila es una lista con 0 (asiento libre) o 1 (asiento ocupado).
print(cine)
```

```python
print(cine[0][1]) # Imprime el valor del asiento de la fila 1, columna 2 (0 o 1)
cine[0][1] = 0 # Libero el asiento de la fila 1, columna 2
print(cine) # Veo como está el cine después de liberar el asiento
```

```python
for i in range(len(cine)):
    print(f"fila {i+1}:")
    for j in range(len(cine[i])):
        print(f"  asiento {j+1}: {'ocupado' if cine[i][j] == 1 else 'libre'}")
```

```python
# Ejercicio pag 245 modulo 2
lista = [[10,20], [30,40], [50,60]]
numero_buscado = 90
encontrado = False
for i in range(len(lista)):
    for j in range(len(lista[i])):
        if lista[i][j] == numero_buscado:
            print(f"Encontrado {numero_buscado} en fila {i+1}, columna {j+1}")
            encontrado = True
            break
    if encontrado:
        break
if not encontrado:
    print(f"{numero_buscado} no encontrado en la lista")
```

Quiero pedir al usuario cuantas filas quiere y cuantas columnas por fila y generar números aleatorios para todos los elementos

```python
import random
filas = int(input("Número de filas: "))
columnas = int(input("Número de columnas: "))
# Bloque para generar la matriz con números aleatorios
matriz = []
for i in range(filas):
    fila = []
    for j in range(columnas):
        valor = random.randint(1, 100) # Genera un número aleatorio entre 1 y 100
        fila.append(valor)
    matriz.append(fila)
    
# Bloque para imprimir la matriz
print("Matriz generada:")
for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        print(matriz[i][j], end=' ')
    print() # Salto de línea después de cada fila
```
