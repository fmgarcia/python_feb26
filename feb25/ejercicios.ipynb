```python
numeros = (1, 2, 5, 4, 3, 2, 1, 4, 7, 8, 9, 9, 3, 7, 3, 9)
frecuencia = {} 
for num in numeros: 
    frecuencia[num] = frecuencia.get(num, 0) + 1  
print(frecuencia)
max_count = max(frecuencia.values())  
result = max(num for num, count in frecuencia.items() if count == max_count)  # encuentra el número con la frecuencia máxima más alto
print(f"El elemento más frecuente es: {result}")
```

```python
tupla = (1, 2, 5, 4, 3, 2, 1, 4, 7, 8, 9, 9, 3, 7, 3, 9)
valor_repetido = max(tupla,  key=lambda x: (tupla.count(x), x))
print(f"El valor que más se repite en la tupla {tupla} es = {valor_repetido}")
```

```python
from collections import Counter
tupla = (1, 2, 5, 4, 3, 1, 1, 4, 7, 8, 9, 9, 3, 7, 3, 9)
conteo = Counter(tupla)
print(conteo)
elementos_comun_mayor = max([num for num, count in conteo.items() if count == conteo.most_common(1)[0][1]])
print(f"El elemento más común es: {elementos_comun_mayor}")
```
