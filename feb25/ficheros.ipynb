```python
# Abrir el fichero en modo lectura
with open('notas.csv', 'r') as fichero:
    # Leer el contenido del fichero
    contenido = fichero.readlines() # Lee todo el fichero y lo guarda en un string


# Imprimir el contenido
print(contenido)
alumnos = contenido[1:] # Elimina la primera línea (cabecera)
print(alumnos)
diccionario_alumnos = {}
for alumno in alumnos:
    nombre, matematicas,lengua,historia = alumno.strip().split(',') # Elimina los espacios y separa por comas
    diccionario_alumnos[nombre] = {
        "matematicas": float(matematicas),
        "lengua": float(lengua),
        "historia": float(historia)
    }
print(diccionario_alumnos)
print(diccionario_alumnos["Fran"])
print(diccionario_alumnos["Fran"]["matematicas"])
for alumno in diccionario_alumnos:
    print(f"{alumno}:")
    for asignatura in diccionario_alumnos[alumno]:
        print(f"  {asignatura}: {diccionario_alumnos[alumno][asignatura]}")
```

### Escribir en un fichero

```python
import random
with open('notas2.csv', 'w') as fichero:
    fichero.write("nombre,matematicas,lengua,historia\n")
    nombres = ["Fran", "Ana", "Luis"]
    for nombre in nombres:
        matematicas = random.randint(0, 10)
        lengua = random.randint(0, 10)
        historia = random.randint(0, 10)
        linea = f"{nombre},{matematicas},{lengua},{historia}\n"
        fichero.write(linea)
```

```python
### Añadir líneas al final de un fichero
with open('notas2.csv', 'a') as fichero:
    nombres = ["María", "Pedro", "Sofía"]
    for nombre in nombres:
        matematicas = random.randint(0, 10)
        lengua = random.randint(0, 10)
        historia = random.randint(0, 10)
        linea = f"{nombre},{matematicas},{lengua},{historia}\n"
        fichero.write(linea)
```

```python
import random
# Importación
from faker import Faker
# Crear instancia (por defecto en inglés)
fake = Faker()
# Crear instancia en español
fake = Faker('es_ES')

numero_personas = 100

with open('notas3.csv', 'w') as fichero:
    fichero.write("nombre,matematicas,lengua,historia\n")
    nombres = [fake.first_name() for _ in range(numero_personas)]
    for nombre in nombres:
        matematicas = random.randint(0, 10)
        lengua = random.randint(0, 10)
        historia = random.randint(0, 10)
        linea = f"{nombre},{matematicas},{lengua},{historia}\n"
        fichero.write(linea)
```
