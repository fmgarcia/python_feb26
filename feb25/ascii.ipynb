```python
# Q2 ejercicio 194 módulo 2
# calcular dias de ventas disminuidos sobre el dia anterior
sales = (100,121,120,130,140,120,122,123,190,125)
decreased_days = [ i+1 for i in range(1, len(sales)) if sales [i] < sales [i-1] ]
print(f"Dias con ventas disminuidas: {decreased_days}, cantidad: {len(decreased_days)}")
```

```python
sales = (100,121,120,130,140,120,122,123,190,125)
contador = 0
for i in range(1, len(sales)):
    if sales [i] < sales [i-1]:
        print(f"Dia {i+1} con ventas disminuidas: {sales[i]} < {sales[i-1]}")
        contador += 1
print(f"Total de dias con ventas disminuidas: {contador}")
```
