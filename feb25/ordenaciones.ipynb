### Diferencia entre sort y sorted

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam']
cadenas.sort() # ordena la lista de cadenas alfabéticamente, modificando la lista original
print (cadenas)
```

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam']
cadenas2 = sorted(cadenas) # ordena la lista de cadenas alfabéticamente, devolviendo una nueva lista ordenada
print(cadenas)
print(cadenas2)
```

### Tipos de orden

#### Orden estándar (tiene en cuenta mayúsculas, minúsculas, acentos, la ñ va por otro lado)

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol']
cadenas.sort()
print(cadenas)
cadenas.sort(reverse=True)
print(cadenas)
```

#### Orden sin tener en cuenta mayúsculas y minúsculas

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol']
cadenas.sort(key=str.lower) # ordena la lista de cadenas alfabéticamente, ignorando mayúsculas y minúsculas
print(cadenas)
```

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol']
cadenas.sort(key=str.casefold) # ordena la lista de cadenas alfabéticamente, ignorando mayúsculas y minúsculas
print(cadenas)
```

#### Orden por número de caracteres

```python

cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol']
cadenas.sort(key=len) # ordena la lista de cadenas por su longitud (número de caracteres)
print(cadenas)
```

#### Orden en español

```python
import locale
# Establecer la configuración regional a español de España
# En Windows suele ser "es_ES" o "Spanish_Spain", en Linux/Mac suele ser "es_ES.UTF-8"
try:
    locale.setlocale(locale.LC_ALL, 'es_ES.UTF-8') # para Linux
except:
    locale.setlocale(locale.LC_ALL, 'Spanish_Spain') # establece la configuración regional a español de España
    
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol', 'Zorro', 'zorro']
cadenas.sort(key=locale.strxfrm) # ordena la lista de cadenas alfabéticamente según las reglas de la configuración regional establecida
print(cadenas)
```

#### Ordenar con múltiples criterios: primero por longitud, luego alfabéticamente (ignorando mayúsculas y minúsculas)

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol', 'Zorro', 'zorro']
cadenas.sort(key=lambda e: (len(e), locale.strxfrm(e))) # ordena la lista de cadenas por su longitud y luego alfabéticamente, ignorando mayúsculas y minúsculas
print(cadenas)
```

#### Orden por la segunda letra de cada cadena, ignorando mayúsculas y minúsculas

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol', 'Zorro', 'zorro']
cadenas.sort(key=lambda e: e[1].lower()) # ordena la lista de cadenas por el segundo carácter de cada cadena, ignorando mayúsculas y minúsculas
print(cadenas)
```

#### Ordenaciones con exclusiones

```python
cadenas = ['Hola', 'Adiós', 'Hasta luego', 'fran', 'Fran', 'Ñam', 'ñam', 'Árbol', 'árbol', 'Zorro', 'zorro', 'Z']
cadenas.sort(key=lambda e: locale.strxfrm(e) if len(e) > 2 else '') # ordena la lista de cadenas por el segundo carácter de cada cadena, ignorando mayúsculas y minúsculas, y maneja el caso de cadenas con un solo carácter
print(cadenas)
```
