```python
for i in range(1, 11): # itera desde i=1 hasta i=10 (inclusive)
    print(i) # imprime el valor actual de i en cada iteración, mostrando los números del 1 al 10 en la salida
```

```python
for i in range(10): # itera desde i=0 hasta i=9 (inclusive)
    print("Hola mundo") # imprime "Hola mundo" en cada iteración, mostrando el mensaje 10 veces en la salida
```

```python
for _ in range(5): # itera 5 veces, utilizando un guion bajo (_) como variable de iteración para indicar que no se necesita su valor
    print("Python es genial") # imprime "Python es genial" en cada iteración, mostrando el mensaje 5 veces en la salida
```

```python
for i in range(1,20,2): # itera desde i=1 hasta i=19 (inclusive) con un paso de 2
    print(i) # imprime el valor actual de i en cada iteración, mostrando los números impares del 1 al 19 en la salida
```

```python
for i in range(10, 0, -1): # itera desde i=10 hasta i=1 (inclusive) con un paso de -1
    print(i) # imprime el valor actual de i en cada iteración, mostrando los números del 10 al 1 en la salida
```
