```python
dia = input("¿En qué día de la semana estamos? ") # se solicita al usuario que introduzca el día de la semana, y se almacena en la variable "dia"
match dia: # se utiliza la estructura de control "match" para comparar el valor de la variable "dia" con diferentes casos
    case "lunes": # si el valor de "dia" es igual a "lunes",
        print("Hoy es lunes, ¡comienza la semana!") # se imprime un mensaje indicando que hoy es lunes y que comienza la semana
    case "martes": # si el valor de "dia" es igual a "martes",
        print("Hoy es martes, ¡a seguir adelante!") # se imprime un mensaje indicando que hoy es martes y que hay que seguir adelante
    case "miércoles": # si el valor de "dia" es igual a "miércoles",
        print("Hoy es miércoles, ¡ya estamos a mitad de semana!") # se imprime un mensaje indicando que hoy es miércoles y que ya se ha llegado a la mitad de la semana
    case "jueves": # si el valor de "dia" es igual a "jueves",
        print("Hoy es jueves, ¡casi es fin de semana!") # se imprime un mensaje indicando que hoy es jueves y que el fin de semana está cerca
    case "viernes": # si el valor de "dia" es igual a "viernes",
        print("Hoy es viernes, ¡por fin es fin de semana!") # se imprime un mensaje indicando que hoy es viernes y que por fin ha llegado el fin de semana
    case "sábado": # si el valor de "dia" es igual a "sábado",  
        print("Hoy es sábado, ¡disfruta tu día de descanso!") # se imprime un mensaje indicando que hoy es sábado y que se debe disfrutar el día de descanso
    case "domingo": # si el valor de "dia" es igual a "domingo",
        print("Hoy es domingo, ¡prepárate para la semana que viene!") # se imprime un mensaje indicando que hoy es domingo y que se debe preparar para la semana que viene
    case _: # si el valor de "dia" no coincide con ninguno de los casos anteriores
        print("¡Día no reconocido! Por favor, introduce un día válido.") # se imprime un mensaje indicando que el día no es reconocido y se solicita al usuario que introduzca un día válido
```

```python
coordenadas = (3, 4) # se define una tupla llamada "coordenadas" con los valores 3 y 4, representando las coordenadas en un plano cartesiano
match coordenadas: # se utiliza la estructura de control "match" para comparar el valor de la tupla "coordenadas" con diferentes casos
    case (0, 0): print("El punto está en el origen") # si las coordenadas son (0, 0), se imprime un mensaje indicando que el punto está en el origen
    case (x, 0): print(f"El punto está sobre el eje X en x={x}") # si las coordenadas son (x, 0), se imprime un mensaje indicando que el punto está sobre el eje X en la posición x
    case (0, y): print(f"El punto está sobre el eje Y en y={y}") # si las coordenadas son (0, y), se imprime un mensaje indicando que el punto está sobre el eje Y en la posición y
    case (x, y) if x > 0 and y > 0: print(f"El punto está en el primer cuadrante en las coordenadas ({x}, {y})") # si las coordenadas son (x, y) con x e y positivos, se imprime un mensaje indicando que el punto está en el primer cuadrante
    case (x, y) if x < 0 and y > 0: print(f"El punto está en el segundo cuadrante en las coordenadas ({x}, {y})") # si las coordenadas son (x, y) con x negativo e y positivo, se imprime un mensaje indicando que el punto está en el segundo cuadrante
    case (x, y) if x < 0 and y < 0: print(f"El punto está en el tercer cuadrante en las coordenadas ({x}, {y})") # si las coordenadas son (x, y) con x e y negativos, se imprime un mensaje indicando que el punto está en el tercer cuadrante
    case (x, y) if x > 0 and y < 0: print(f"El punto está en el cuarto cuadrante en las coordenadas ({x}, {y})") # si las coordenadas son (x, y) con x positivo e y negativo, se imprime un mensaje indicando que el punto está en el cuarto cuadrante
```
