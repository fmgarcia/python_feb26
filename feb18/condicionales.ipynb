```python
"""
Ejercicio pag 344
Recibiremos un número por pantalla entre -100 y 100.
Imprimimos el número en formato x = numero
Si el número es positivo, imprimimos 'El número es positivo'
"""
numero = int(input("Introduce un número entre -100 y 100: "))
print(f"x = {numero}") # imprime el número introducido por el usuario en formato "x = numero", lo cual muestra el valor de la variable "numero" que contiene el número introducido por el usuario
if numero > 0:
    print("El número es positivo") # imprime un mensaje si la condición numero > 0 es verdadera, lo cual indica que el número introducido por el usuario es positivo
```

```python
# Ejemplos de uso de split
frase = "Hola, ¿cómo estás?"
palabras = frase.split() # se utiliza el método split() para dividir la cadena de texto "frase" en una lista de palabras, utilizando el espacio como separador por defecto
print(palabras) # imprime la lista de palabras obtenida al dividir la frase, lo cual muestra ['Hola,', '¿cómo', 'estás?'] como resultado, indicando que la frase se ha dividido correctamente en palabras individuales
```

```python
# ejemplo de uso de split con un separador específico
fecha = "2024-06-01"
partes_fecha = fecha.split("-") # se utiliza el método split() para dividir la cadena de texto "fecha" en una lista de partes, utilizando el guion "-" como separador
print(partes_fecha) # imprime la lista de partes obtenida al dividir la fecha, lo cual muestra ['2024', '06', '01'] como resultado, indicando que la fecha se ha dividido correctamente en año, mes y día utilizando el guion como separador
```

```python
# Ejemplo de splitlines
texto_multilinea = """Hola, ¿cómo estás?
Espero que estés bien.
Saludos.
"""
lineas = texto_multilinea.splitlines() # se utiliza el método splitlines() para dividir la cadena de texto "texto_multilinea" en una lista de líneas, utilizando el salto de línea como separador
print(lineas) # imprime la lista de líneas obtenida al dividir el texto multilinea, lo cual muestra ['', 'Hola, ¿cómo estás?', 'Espero que estés bien.', 'Saludos.', ''] como resultado, indicando que el texto se ha dividido correctamente en líneas individuales, incluyendo las líneas vacías al principio y al final debido a los saltos de línea
```

```python
""" Ejercicio pag 357
Recibiremos una edad por pantalla, si la edad es 20 o más imprimiremos 'Eres un hombre adulto', 
si la edad es menor a 20 pero mayor o igual a 10 imprimiremos 'Eres un hombre joven', 
si la edad es menor a 10 imprimiremos 'Eres un niño'
"""
edad = int(input("Introduce tu edad: "))
if edad >= 20:
    print("Eres un hombre adulto") # imprime un mensaje si la condición edad >= 20 es verdadera, lo cual indica que la persona es un hombre adulto
elif edad >= 10:
    print("Eres un hombre joven") # imprime un mensaje si la condición edad >= 10 es verdadera, lo cual indica que la persona es un hombre joven
else:
    print("Eres un niño") # imprime un mensaje si ninguna de las condiciones anteriores es verdadera, lo cual indica que la persona es un niño
```

```python
numero = 15
if numero > 0:
    print("El número es positivo") # imprime un mensaje si la condición numero > 0 es verdadera, lo cual indica que el número es positivo
elif numero < 0:
    print("El número es negativo") # imprime un mensaje si la condición numero < 0 es verdadera, lo cual indica que el número es negativo
else:
    print("El número es cero") # imprime un mensaje si ninguna de las condiciones anteriores es verdadera, lo cual indica que el número es cero
```

```python
import random
aleatorio = random.randint(1,3) # se utiliza la función randint() del módulo random para generar un número entero aleatorio entre 1 y 3, y se asigna a la variable "aleatorio"
if aleatorio == 1:
    portero = "izquierda"
elif aleatorio == 2:
    portero = "centro"
else:
    portero = "derecha"
jugador = input("¿A qué lado quieres tirar? (izquierda, centro, derecha): ") # se solicita al usuario que introduzca un lado para tirar el balón, y se almacena en la variable "jugador"
if jugador == portero:
    print("¡El portero ha parado el balón!") # imprime un mensaje si la elección del jugador coincide con la posición del portero, lo cual indica que el portero ha detenido el balón
else:
    print("¡Has marcado un gol!") # imprime un mensaje si la elección del jugador no coincide con la posición del portero, lo cual indica que el jugador ha marcado un gol
```
